### L'algorithme « donné sur UMTICE » ([`algo_pivot_de_gauss.pdf`](algo_pivot_de_gauss.pdf))

*Recopié ici depuis le PDF rangé dans ce dossier ; les indices commencent à 1, comme dans le cours.*

**Algorithme 1 — Pivot de Gauss**

**Entrées :** $m, n \in \mathbb{N}$ ; $A \in \mathcal{M}_{m,n}(\mathbb{R})$
**Sorties :** $A'$ forme échelonnée réduite de $A$ ; $P = (p_h)$ la liste des pivots

1. $r \leftarrow 0$ *// indice de la ligne du dernier pivot*
2. **Pour** $j$ de 1 jusqu'à $n$ **Faire** *// $j$ décrit tous les indices de colonne*
3. &emsp;Chercher la ligne $i > r$ où $|a_{i,j}|$ est maximal et non nul
4. &emsp;**Si** cette ligne $i$ existe **Alors**
5. &emsp;&emsp;$r \leftarrow r + 1$
6. &emsp;&emsp;Permuter la ligne $i$ et la ligne $r$
7. &emsp;&emsp;Ajouter le pivot $a_{r,j}$ à la liste
8. &emsp;&emsp;$L_r \leftarrow L_r / a_{r,j}$ *// Normalisation du pivot*
9. &emsp;&emsp;**Pour** $i$ de 1 jusqu'à $n$ **Faire** *// Elimination dans les autres lignes*
10. &emsp;&emsp;&emsp;**Si** $i \neq r$ **Alors**
11. &emsp;&emsp;&emsp;&emsp;$L_i \leftarrow L_i - a_{i,j} L_r$ *// On annule $a_{i,j}$*
12. &emsp;&emsp;&emsp;**Fin Si**
13. &emsp;&emsp;**Fin Pour**
14. &emsp;**Sinon**
15. &emsp;&emsp;Ajouter le pivot 0 à la liste
16. &emsp;**Fin Si**
17. **Fin Pour**
