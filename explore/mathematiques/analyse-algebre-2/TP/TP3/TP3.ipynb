## II Méthode des rectangles

Pour une fonction à valeurs positives, la valeur d'une intégrale sur un intervalle est l'aire de la partie du plan située entre la courbe et l'axe des abscisses.

Une méthode pour approcher cette aire est de construire des rectangles dont le côté droit va de $(x, 0)$ à $(x, f(x))$. En sommant les aires de ces rectangle, on obtient une valeur proche de celle de l'intégrale.

## IV Monte Carlo

Cette méthode se base sur des théorèmes de probabilités pour estimer la valeur de l'intégrale. D'une fois sur l'autre, le résultat ne sera donc pas exactement le même.

L'idée est de considérer un rectangle d'aire $R$ dans lequel il est possible de dessiner le morceau de courbe où l'on veut calculer l'intégrale. La courbe sépare le rectangle en deux parties : la partie correspondant à l'aire $A$ que l'on veut calculer, et l'autre partie.

On va « lancer » un certain nombre de points au hasard (répartition uniforme) dans le rectangle. La probabilité que le point tombe sous la courbe est égale à la $\frac{A}{R}$ car c'est la proportion du rectangle qui correspond à cette partie. Ainsi, en lançant un grand nombre de points, la loi des grands nombres nous dit que la proportion de points sous la courbe tend vers la probabilité $\frac{A}{R}$. Connaissant $R$, il est donc possible de retrouver la valeur $A$ de notre intégrale.

![Méthode de Monte-Carlo (figure de l'énoncé)](figures/monte-carlo.png)
