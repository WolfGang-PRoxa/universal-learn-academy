## II Recherche de zéros par dichotomie

La méthode de dichotomie consiste, sur un ensemble de données ordonné, à rechercher une valeur $x_0$ particulière se trouvant entre deux valeurs données $a$ et $b$. A chaque étape, on sépare l'intervalle de recherche en deux en calculant le milieu $c$. Si $x_0$ se trouve entre $a$ et $c$, on remplace $b$ par $c$. Sinon, on remplace $a$ par $c$.

Appliquons cela à une fonction continue $f$ dont on connait deux réels $a$ et $b$ pour lesquels $f(a)f(b) < 0$ (ce qui signifie qu'ils sont de signe différents). La fonction étant continue, d'après le théorème des valeurs intermédiaires, il existe un réel $x_0$ compris entre $a$ et $b$ tel que $f(x_0) = 0$.

Écrire une fonction `ZeroDichotomie(f, xMin, xMax, nbIter, tracer = False)` qui effectue `nbIter` itérations où l'intervalle de recherche est coupé en deux. Cette fonction renvoie une valeur approchée du zéro de la fonction.

Lorsque le paramètre `tracer` est vrai, on souhaite tracer à chaque itération un petit trait vertical rouge indiquant le milieu de l'intervalle d'étude.

## III Méthode de Newton

Si la fonction a un comportement « assez régulier », il est possible d'utiliser une méthode bien plus efficace : la méthode de Newton. Elle s'approche du zéro en beaucoup moins d'itérations.

Cette méthode est itérative. Sur l'exemple dessiné ci-dessous, le point de départ est l'extrémité droite de l'intervalle. En ce point $x$, on calcule la dérivée de $f$. On peut alors tracer la tangente (équation de la forme $y = ax + b$) et déterminer où la tangente croise l'axe des abscisse. Ce point sera (sauf situation particulières) plus proche du zéro que le point de départ. On recommence en prenant ce point comme $x$.
