### Norme d'un vecteur

Exemples du cours avec $x = \begin{bmatrix} 2\\ 3 \\ -1 \end{bmatrix}$ et  $y = \begin{bmatrix} 2 \\ 3 \\ 10 \end{bmatrix}$. Calculer les normes, $L_1$ et $L_2$

```python
def norme_l1(a):
    return np.sum(np.abs(a))

def norme_l2(a):
    return np.sqrt(np.sum(a*a))

x = np.array([2,3,-1])
y = np.array([2, 3, 10])

print(norme_l1(x), norme_l2(x), norme_l1(y), norme_l2(y))
```

$\Vert x \Vert_1 = 6$, 
$\Vert x \Vert_2 = 3.74$,
$\Vert x \Vert_1 = 15$,
$\Vert x \Vert_2 = 10.63$

### Inverse d'une matrice

CM1 p44

$ A = \begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix}$ et 
$B = \begin{bmatrix} -1 & 1 & 0 \\ 3 & 2 & 5 \\ 0 & 2 & 2 \end{bmatrix}$

```python
def calcul_determinant(mat):
    #mat : matrice (2,2)
    return mat[0,0]*mat[1,1]-mat[0,1]*mat[1,0]
A = np.array([[1, 2],[3, 4]])
detA= calcul_determinant(A)
print(detA)

invA = -1/2 * np.array([[4, -2],[-3,1]]) #trouvé en cours
# vérification A invA = I2
print(invA.dot(A))
```

```python
B = np.array([[-1, 1, 0],[3, 2, 5],[0, 2, 2]])
detB= np.linalg.det(B)
print(detB)
```
